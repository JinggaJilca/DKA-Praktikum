### Program Diskon

```python
def hitungDiskon(hargaAwal, persentase):
    potongan = (hargaAwal * persentase)/100
    return hargaAwal - potongan

harga = int(input("Masukkan harga awal: "))
persen = int(input("Masukkan persentase: "))
hasilDiskon = hitungDiskon(harga, persen)
print(f"Hasil Diskon = Rp{hasilDiskon:,.0f}")
```

### NetworkX

```python
import networkx as nx

#Fungsi cetak graf
def showGraf(G, pos, title=""):
    if pos == None:
        pos = nx.spring_layout(G)

    nx.draw(
        G,
        pos,
        with_labels=True,
        node_color="red",
        node_size=200,
        font_color="white",
        font_weight="bold",
        width=5)
    
    edge_labels = nx.get_edge_attributes(G,'weight')
    nx.draw_networkx_edge_labels(G, pos=None, edge_labels=edge_labels, font_color="black")
    plt.margins(0.5)
    plt.figure(figsize=(10,5))
    plt.title(title)
    plt.show()

# Buat Graf
G = nx.Graph()

# Edges
edge_weight_data = [
    ('Jakarta', 'Cirebon', 0), 
    ('Jakarta','Bandung', 0),
    ('Cirebon','Bandung', 0),
    ('Cirebon','Semarang', 0),
    ('Cirebon', 'Yogyakarta', 0),
    ('Bandung','Yogyakarta', 0),
    ('Semarang','Surabaya', 0),
    ('Semarang','Surakarta', 0),
    ('Semarang','Yogyakarta', 0),
    ('Yogyakarta','Surakarta', 0),
    ('Surakarta','Malang', 0),
    ('Malang','Surabaya', 0)
]
G.add_weighted_edges_from(edge_weight_data)

showGraf(G,pos=None, title="Graph Kosong")
```

### BFS

```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.Graph()

# Posisi
pos = {
    'A': (0, 3), 
    'B': (-10, 2),
    'C': (10, 2), 
    'D': (-15, 1),
    'E': (5, 1)
}
print("Posisi D: ", pos.get('D'))

edges = [
    ('A','B'),('A', 'C'),
    ('B','D'),
    ('C',"E")
]

# Nambahin edges aja
G.add_edges_from(edges)

# Buat nambahin weight ke G
# G.add_weighted_edges_from(edges)

print(G.nodes())
print(G.edges())
print(G.number_of_edges())
print(G.number_of_nodes())

#Set ukuran framenya
plt.figure(figsize=(12, 8))

# Set Node
nx.draw(
    G,
    with_labels=True,
    pos=pos,
    width=3,
    node_color="blue",
    node_shape="o",
    font_weight="bold",
    font_color="white",
    font_size= 15,
    node_size= 2500
)

edge_labels = nx.get_edge_attributes(G, 'weight')

# Set Label/weightnya
nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=edge_labels,
    font_color="red",
    font_weight="bold",
    font_size=12
)


plt.title("First try")
plt.axis('off')
plt.margins(0.2)
plt.show()


# Mengunjungi node
print("Urutan node BFS:")
i = 1
for nodes in nx.bfs_edges(G, source="A"):
    print(f"Ke {i} : {nodes[0]} -> {nodes[1]}")
    i+=1

print("-"*20)
print("Urutan node DFS: ")
i=1
for nodes in nx.dfs_edges(G, source="A"):
    print(f"Ke {i} : {nodes[0]} -> {nodes[1]}")
    i += 1

```
