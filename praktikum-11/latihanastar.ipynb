### Soal 1

```python
import networkx as nx
import matplotlib.pyplot as plt
import math
 
#Membuat wadah kosong untuk graph undirected
G_undirected = nx.Graph()

# Inisialisasi Weight tiap node
edge_weight_data = [
    ('Arad', 'Zerind', 75),
    ('Zerind', 'Oradea', 71),
    ('Oradea', 'Sibiu', 151),
    ('Arad', 'Sibiu', 140),
    ('Arad', 'Timisoara', 118),
    ('Timisoara', 'Lugoj', 111),
    ('Lugoj', 'Mehadia', 70),
    ('Mehadia', 'Drobeta', 75),
    ('Drobeta', 'Craiova', 120),
    ('Craiova', 'Pitesti', 138),
    ('Craiova', 'Rimnicu Vilcea', 146),
    ('Sibiu', 'Rimnicu Vilcea', 80),
    ('Sibiu', 'Fagaras', 99),
    ('Rimnicu Vilcea', 'Pitesti', 97),
    ('Fagaras', 'Bucharest', 211),
    ('Pitesti', 'Bucharest', 101),
    ('Bucharest', 'Giurgiu', 90),
    ('Bucharest', 'Urziceni', 85),
    ('Urziceni', 'Hirsova', 98),
    ('Hirsova', 'Eforie', 86),
    ('Urziceni', 'Vaslui', 142),
    ('Vaslui', 'Iasi', 92),
    ('Iasi', 'Neamt', 87)
]
 
#Membuat daftar untuk agar saling terhubung
G_undirected.add_weighted_edges_from(edge_weight_data)
 
#Menentukan posisi sesuai gambar menggunakan koordinat
pos = {
    'Oradea': (1.5, 10), 
    'Zerind': (1, 8.5), 
    'Arad': (0.5, 7),
    'Timisoara': (0.5, 4.5), 
    'Lugoj': (1.5, 3.5), 
    'Mehadia': (1.5, 2.3),
    'Drobeta': (1.5, 1), 
    'Sibiu': (2, 6), 
    'Rimnicu Vilcea': (3.5, 5),
    'Craiova': (4, 1), 
    'Fagaras': (5, 6), 
    'Pitesti': (5.5, 4),
    'Bucharest': (7, 2.5), 
    'Giurgiu': (6.5, 1), 
    'Urziceni': (8, 3),
    'Vaslui': (9, 6), 
    'Iasi': (8.5, 8), 
    'Neamt': (7.5, 9),
    'Hirsova': (9.5, 3), 
    'Eforie': (10, 1.5)
}
 
#Set ukuran framenya
plt.figure(figsize=(12, 8))
 
#Menggambar node
nx.draw(G_undirected, pos, 
            with_labels=True, 
            node_color='lightgrey', 
            node_shape='s',
            node_size=2500,
            font_size=8,
            font_weight='bold',
            edge_color='black',
            linewidths=2)
 
#Data weight
edge_labels = nx.get_edge_attributes(G_undirected, 'weight')
 
#Tulisan bobot pada edges
nx.draw_networkx_edge_labels(G_undirected, pos, edge_labels=edge_labels, font_size=8)
 
plt.title("Latihan GBFS Arad - Bucharest")
plt.axis('off')
plt.show()
 
 
# Fungsi Heuristik menggunakan Euclidean Distance
def heuristable(u, v):

    x1, y1 = pos[u]
    x2, y2 = pos[v]
    
    # Rumus Euclidean: akar dari (delta x^2 + delta y^2)
    return math.sqrt((x2 - x1)**2 + (y2 - y1)**2)


path_node = nx.astar_path(
    G_undirected, 
    "Arad", 
    "Bucharest", 
    heuristic=heuristable, 
    weight='weight'
) 

# Cetak hasil
print("Node yang dikunjungi dari Arad ke Bucharest: ", path_node)

for node in path_node:
    # Menghitung jarak dari node saat ini ke tujuan (Bucharest)
    h_value = heuristable(node, "Bucharest")
    
    if node == "Bucharest":
        print(f"[{node}] -> TIBA DI TUJUAN (Heuristik: {h_value:.2f})")
    else:
        print(f"[{node}] -> Heuristik ke {"Bucharest"}: {h_value:.2f}")
 
```
