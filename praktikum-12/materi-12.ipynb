## Pendefinisian Simbol Pada SymPy 

```python
import sympy as sp
```

```python
# Penggunaan symbols() untuk mendefinisikan satu simbol 
x = sp.symbols('x') 
x
```

```python
# Pendefinisian beberapa simbol  
# Cara 1 
a, b, c = sp.symbols("a b c") 
# Cara 2 
a, b, c = sp.symbols("a, b, c") 
a + b + c
```

```python
# Pendefinisian simbol dengan list 
my_symbols = ['a', 'b', 'c'] 
a, b, c = sp.symbols(my_symbols) 
a + b + c 
```

```python
# Pendefinisian simbol variabel dengan indeks 
x1, x2, x3 = sp.symbols("x1, x2, x3") 
x1 + x2 + x3
```

```python
# Pendefinisian simbol dengan notasi range 
x_n = sp.symbols('x:5') 
x_n[0] + x_n[1] + x_n[2] + x_n[3] + x_n[4]
```

```python
# Pendefinisian simbol Yunani 
alpha, beta, gamma = sp.symbols("alpha, beta, gamma") 
alpha + beta + gamma 
```

## Aljabar Dasar Pada SymPy 

```python
# Penjumlahan 
x, y = sp.symbols('x, y') 
addition = x + y 
addition 
```

```python
# Pengurangan 
x, y = sp.symbols('x, y') 
substraction = x - y 
substraction 
```

```python
# Perkalian 
x, y = sp.symbols('x, y') 
multiplication = x * y 
multiplication
```

```python
# Pembagian 
x, y = sp.symbols('x, y') 
division = x / y 
division 
```

```python
x, y = sp.symbols('x, y') 
expr = (x + y) ** 2 
expr 
```

```python
# Mengembangkan ekspresi 
expanded_expr = expr.expand() 
expanded_expr 
```

```python
# Simplicity
x = sp.symbols('x') 
expr = sp.sin(x)**2 + sp.cos(x)**2  
expr 
```

```python
# Menyederhanakan ekspresi 
simplified_expr = expr.simplify() 
simplified_expr 
```

## Kalkulus Dasar

```python
x = sp.symbols('x') 
# Pendefinisian fungsi f(x) 
f = x**3 + 2*x**2 + x 
f 
```

```python
# Difrensiasi (Turunan) fungsi f(x) terhadap x 
f_prime = f.diff(x) 
f_prime 
```

```python
x = sp.symbols('x') 
# Pendefinisian fungsi f(x) 
f = sp.sin(x) 
f
```

```python
# Integrasi tak tentu 
f_integral_indef = sp.integrate(f, x) 
f_integral_indef
```

```python
# Integrasi tentu dari 0 sampai pi 
f_integral_def = sp.integrate(f, (x, 0, sp.pi)) 
# .integrate(fungsi, (ekspresi, batas_bawah, batas_atas)) 
f_integral_def
```

```python
x = sp.symbols('x') 
# Pendefinisian fungsi f(x) 
f = sp.sin(x) / x 
f 
```

```python
# Limit
# Menghitung limit f(x) saat x mendekati 0 
f_limit = sp.limit(f, x, 0)  
f_limit 
```

## Persamaan Simbolik

```python
x = sp.symbols('x') 
# Membuat persamaan 
equation = sp.Eq(x**2 - 4, 0) 
equation 
```

```python
x, y = sp.symbols('x, y') 
# Membuat persamaan 
equation = sp.Eq(x**3 - 2, y - 2 + 3*x) 
equation 
```

```python
x = sp.symbols('x') 
# Membuat persamaan 
equation = sp.Eq(x**2 - 4, 0) 
equation 
```

```python
# Memecahkan persamaan untuk x 
solution = sp.solve(equation, x) 
solution
```

```python
# Pendefinisian simbol dan fungsi 
t = sp.symbols('t') 
y = sp.Function('y')
 
equation = sp.Eq(y(t).diff(t), -t + y(t)) 
equation
```

```python
# Penyelesaian persamaan difrensial 
solution = sp.dsolve(equation, y(t)) 
solution
```
