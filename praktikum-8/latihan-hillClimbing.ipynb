```python
import random

# Representasi jarak antar kota (berdasarkan tabel)
cities = ['A', 'B', 'C', 'D', 'E']

distance_matrix = {
    ('A','A'):0, ('A','B'):10, ('A','C'):15, ('A','D'):20, ('A','E'):25,
    ('B','A'):10, ('B','B'):0, ('B','C'):35, ('B','D'):25, ('B','E'):15,
    ('C','A'):15, ('C','B'):35, ('C','C'):0, ('C','D'):30, ('C','E'):20,
    ('D','A'):20, ('D','B'):25, ('D','C'):30, ('D','D'):0, ('D','E'):10,
    ('E','A'):25, ('E','B'):15, ('E','C'):20, ('E','D'):10, ('E','E'):0
}

# Menghitung total jarak rute
def total_distance(tour):
    dist = 0
    for i in range(len(tour)):
        dist += distance_matrix[(tour[i], tour[(i + 1) % len(tour)])]
    return dist

# Generate neighbors (swap 2 kota)
def get_neighbors(tour):
    neighbors = []
    for i in range(len(tour)):
        for j in range(i + 1, len(tour)):
            neighbor = tour[:]
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
            neighbors.append(neighbor)
    return neighbors


# Solusi awal (acak)
current_solution = cities[:]
random.shuffle(current_solution)

current_distance = total_distance(current_solution)

print("Solusi awal:", " -> ".join(current_solution + [current_solution[0]]))
print("Jarak awal:", current_distance)
print("-"*30)

iteration = 1

while True:
    neighbors = get_neighbors(current_solution)

    # Cari neighbor terbaik
    best_neighbor = min(neighbors, key=lambda t: total_distance(t))
    best_distance = total_distance(best_neighbor)

    print(f"Iterasi {iteration}:")
    print("Best neighbor:", best_neighbor)
    print("Distance:", best_distance)
    print("-"*30)

    # Jika tidak lebih baik → berhenti
    if best_distance >= current_distance:
        break

    iteration += 1

    # Update solusi
    current_solution = best_neighbor
    current_distance = best_distance

print("\n=== HASIL AKHIR ===")
print("Rute terbaik:", " -> ".join(current_solution + [current_solution[0]]))
print("Total jarak:", current_distance)
```
