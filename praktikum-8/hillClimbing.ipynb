```python
import random 
import math

# Menghitung jarak 2 titik dengan menggunakan rumus euclidean 
def distance(city1, city2): 
    return math.sqrt((city1[0] - city2[0]) ** 2 + (city1[1] - city2[1]) ** 2) 
 
# Menjumlah seluruh jarak antar titik pada rute tour 
def total_distance(tour, cities): 
    dist = 0 
    for i in range(len(tour)): 
        dist += distance(cities[tour[i]], cities[tour[(i + 1) % len(tour)]]) 
    return dist 
 
# Memberikan semua kombinasi rute yang mungkin dengan cara menukar dua titik dalam rute tour yang diberikan 
def get_neighbors(tour): 
    neighbors = [] 
    for i in range(len(tour)):   
        for j in range(i + 1, len(tour)): 
            neighbor = tour[:] 
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i] 
 
            neighbors.append(neighbor) 
 
    return neighbors 

# Deklarasi seluruh titik kota (x, y) dalam bentuk tuple kedalam list cities 
cities = [ 
    (0, 0), (1, 3), (4, 3), (6, 1), 
    (3, 0), (2, 2), (5, 5), (7, 7) 
] 

current_solution = list(range(len(cities))) # deklarasi rute kota awal 
random.shuffle(current_solution) # mengacak rute awal menggunakan fungsi random 
 
# Menetapkan total jarak untuk rute awal 
current_distance = total_distance(current_solution, cities)  
 
# Program akan terus melakukan loop hingga tidak menemukan nilai distance    
# yang lebih kecil dari current distance 
while True: 
    # Mendapatkan seluruh kombinasi rute dengan cara menukar 2 titik  
    # pada current solution 
    neighbors = get_neighbors(current_solution)  
 
    # Menggunakan fungsi lambda untuk mencari nilai minimum  
    # berdasarkan total distance dalam list neighbors 
    best_neighbor = min(neighbors, key=lambda tour: total_distance(tour,  
cities)) 
 
    # Menghitung total distance berdasarkan best neighbor yang didapat  
    best_distance = total_distance(best_neighbor, cities) 
         
    # Jika tidak ada neighbor yang lebih baik, keluar dari loop 
    if best_distance >= current_distance: 
        break 
         
    # Update solusi saat ini dengan solusi terbaik dari tetangga 
    current_solution = best_neighbor 
    current_distance = best_distance 
  
print("Solusi terbaik (urutan kota):", current_solution) 
print("Total jarak terbaik:", current_distance) 
```
