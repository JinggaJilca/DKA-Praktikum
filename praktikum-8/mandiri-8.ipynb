### Soal 1

```python
import random 
import math

# 2. Fungsi menghitung jarak euclid
def distance(city1, city2): 
    return math.sqrt((city1[0] - city2[0]) ** 2 + (city1[1] - city2[1]) ** 2) 
 
# 3. Fungsi menghitung total jarak
def total_distance(tour, cities): 
    dist = 0 
    for i in range(len(tour)): 
        dist += distance(cities[tour[i]], cities[tour[(i + 1) % len(tour)]]) 
    return dist 
 
# Fungsi untuk mendapatkan neighbors (menukar posisi)
def get_neighbors(tour): 
    neighbors = [] 
    for i in range(len(tour)):   
        for j in range(i + 1, len(tour)): 
            neighbor = tour[:] 
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i] 
            neighbors.append(neighbor) 
    return neighbors 

# Deklarasi seluruh titik kota (x, y)
cities = [ 
    (0, 0), (1, 3), (4, 3), (6, 1), 
    (3, 0), (2, 2), (5, 5), (7, 7) 
] 

# 4. Implementasi Hill Climbing
# Solusi Awal
current_solution = list(range(len(cities)))
random.shuffle(current_solution)
 
# Menetapkan total jarak untuk rute awal 
current_distance = total_distance(current_solution, cities)  
 
# 5. Menampilkan rute awal dan jarak sebelum optimasi
print("Rute Awal: ", current_solution)
print(f"Total jarak sebelum optimasi: {current_distance:.2f}")
print("-" * 50)


# Loop HC
while True:
    # Menukar dua posisi kota (swap)
    neighbors = get_neighbors(current_solution)

    # Seleksi neighbor terbaik
    best_neighbors = min(neighbors, key=lambda tour: total_distance(tour, cities))
    best_distance = total_distance(best_neighbors, cities)

    # 6. Menemukan solusi yg lebih baik
    if best_distance >= current_distance:
        print("Solusi optimal lokal telah ditemukan")
        break
    current_solution = best_neighbors
    current_distance = best_distance

# Menampilkan rute terbaik dan total jarak sesudah optimasi
print("Rute terbaik hasil optimasi:", current_solution) 
print(f"Total jarak sesudah optimasi: {current_distance:.2f}")
```

### Soal 2

```python
import random

# 1. Representasi tugas dalam bentuk list
tasks = ['A', 'B', 'C', 'D', 'E']

# Waktu dasar pengerjaan masing-masing tugas
base_times = {'A': 10, 'B': 15, 'C': 8, 'D': 20, 'E': 12}

transition_times = {
    'A': {'A': 0, 'B': 2, 'C': 5, 'D': 4, 'E': 1},
    'B': {'A': 5, 'B': 0, 'C': 2, 'D': 6, 'E': 4},
    'C': {'A': 4, 'B': 1, 'C': 0, 'D': 3, 'E': 5},
    'D': {'A': 7, 'B': 5, 'C': 2, 'D': 0, 'E': 3},
    'E': {'A': 2, 'B': 4, 'C': 6, 'D': 1, 'E': 0}
}

# 2. Fungsi Evaluasi (Cost Function)
def calculate_cost(sequence):
    total_time = 0
    for i in range(len(sequence)):
        # Menambahkan waktu dasar pengerjaan
        total_time += base_times[sequence[i]]
        
        # Menambahkan waktu transisi jika belum di tugas terakhir
        if i < len(sequence) - 1:
            total_time += transition_times[sequence[i]][sequence[i+1]]
            
    return total_time

# Fungsi untuk mendapatkan neighbor (menukar dua posisi tugas)
def get_neighbors(sequence):
    neighbors = []
    for i in range(len(sequence)):
        for j in range(i + 1, len(sequence)):
            neighbor = sequence[:]
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i] # Terjadi swap
            neighbors.append(neighbor)
    return neighbors

# 3. Implementasi Hill Climbing
current_solution = tasks[:]
random.shuffle(current_solution)
current_cost = calculate_cost(current_solution)

# 5. Menampilkan Solusi Awal
print("=== PROSES PENJADWALAN TUGAS (HILL CLIMBING) ===")
print(f"Solusi Awal: {current_solution} | Cost: {current_cost}")
print("-" * 50)

iteration = 1
# Loop Hill Climbing
while True:
    neighbors = get_neighbors(current_solution)
    
    # Pilih solusi dengan cost lebih kecil
    best_neighbor = min(neighbors, key=calculate_cost)
    best_cost = calculate_cost(best_neighbor)
    
    # 4. Jika neighbor terbaik memiliki cost yang LEBIH KECIL, kita update
    if best_cost < current_cost:
        print(f"Iterasi {iteration}:")
        print(f"  Urutan berubah menjadi : {best_neighbor}")
        print(f"  Cost turun menjadi     : {best_cost}")
        current_solution = best_neighbor
        current_cost = best_cost
        iteration += 1
    else:
        # Jika tidak ada cost yang lebih kecil, algoritma berhenti
        print(f"\nIterasi berhenti")
        break

print("-" * 50)
# 5. Output: Solusi Terbaik dan Total Cost Akhir
print("=== HASIL AKHIR ===")
print(f"Solusi Terbaik    : {current_solution}")
print(f"Total Cost Akhir  : {current_cost}")
```
