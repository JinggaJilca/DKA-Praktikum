Fuzzy Logic adalah pendekatan komputasi yang meniru cara manusia membaut 
keputusan berdasarkan informasi yang tidak pasti atau ambigu. Tidak seperti logika 
tradisional yang hanya mengenal nilai 0 (salah) dan 1 (benar), fuzzy logic 
memperkenalkan konsep derajat kebenaran, di mana nilai bisa berada di antara 0 dan 
1, mencerminkan tingkat kepastian dan ketidakpastian. 

1. Fuzzifikasi mengubah nilai numerik menjadi nilai fuzzy
2. Fuzzy Rules kumpulan aturan jika-maka
3. Inference Engine kompenen mengevaluasi aturan fuzzy bedasarkan input yang telah di fuzzifikasi
4. Defuzzifikasi mengubah nilai fuzzy menjadi nilai numberik sebagai output

```python
# Import module yang diperlukan 
import skfuzzy as fuzz 
import numpy as np 
from skfuzzy import control as ctrl 
import matplotlib.pyplot as plt 

# Definisi varibel input 
quality = ctrl.Antecedent(np.arange(0, 11, 1), 'quality') 
service = ctrl.Antecedent(np.arange(0, 11, 1), 'service') 
# Definisi variabel output 
tip = ctrl.Consequent(np.arange(0, 26, 1), 'tip') 


# Fungsi keanggotaan untuk quality 
quality['poor'] = fuzz.trimf(quality.universe, [0,0,5]) 
quality['average'] = fuzz.trimf(quality.universe, [0,5,10]) 
quality['good'] = fuzz.trimf(quality.universe, [5,10,10]) 
 
# Visualisasi 
quality.view() 

# Fungsi keanggotaan untuk service 
service['poor'] = fuzz.trimf(service.universe, [0, 0, 5]) 
service['average'] = fuzz.trimf(service.universe, [0, 5, 10]) 
service['good'] = fuzz.trimf(service.universe, [5, 10, 10]) 
 
service.view() 

# Fungsi keanggotaan untuk tip 
tip['low'] = fuzz.trimf(tip.universe, [0, 0, 13]) 
tip['medium'] = fuzz.trimf(tip.universe, [0, 13, 25]) 
tip['high'] = fuzz.trimf(tip.universe, [13, 25, 25]) 
 
tip.view() 

# Membuat aturan dari sistem fuzzy 
rule1 = ctrl.Rule(quality['poor'] | service['poor'], tip['low']) 
rule2 = ctrl.Rule(service['average'], tip['medium']) 
rule3 = ctrl.Rule(service['good'] | quality['good'], tip['high'])


# Membuat objek untuk defuzzification 
tipping_ctrl = ctrl.ControlSystem([rule1, rule2, rule3]) 
tipping_simulation = ctrl.ControlSystemSimulation(tipping_ctrl)

# Memberikan input pada variabel Antecedent 
tipping_simulation.input['quality'] = 6.5 
tipping_simulation.input['service'] = 9.8 
# Menghitung output dari sistem fuzzy 
tipping_simulation.compute() 

# Menampilkan hasil & visualisasi perhitungan 
print(tipping_simulation.output['tip'])
tip.view(sim=tipping_simulation) 
```
