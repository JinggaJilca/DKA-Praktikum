```python
# Import module yang diperlukan
import skfuzzy as fuzz
import numpy as np
from skfuzzy import control as ctrl
import matplotlib.pyplot as plt

# Definisi varibel input
quality = ctrl.Antecedent(np.arange(0, 11, 1), 'quality')
service = ctrl.Antecedent(np.arange(0, 11, 1), 'service')
# Definisi variabel output
tip = ctrl.Consequent(np.arange(0, 26, 1), 'tip')

# Fungsi keanggotaan untuk quality
# Menggunakan trapesium (trapmf) = [mulai naik, mulai puncak, akhir puncak, kembali ke dasar]
quality['poor'] = fuzz.trapmf(quality.universe, [0, 0, 2, 5])
quality['average'] = fuzz.trapmf(quality.universe, [0, 4, 6, 10])
quality['good'] = fuzz.trapmf(quality.universe, [5, 8, 10, 10])


# Visualisasi
quality.view()

# Fungsi keanggotaan untuk service
service['poor'] = fuzz.trapmf(quality.universe, [0, 0, 3, 5])
service['average'] = fuzz.trapmf(quality.universe, [2, 4, 8, 10])

# Menggunakan segitiga (trimf) = [mulai naik, puncak, turun]
service['good'] = fuzz.trimf(quality.universe, [5,10, 10])

service.view()
```
