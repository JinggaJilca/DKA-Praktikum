## Soal 1: Implementasi Hill Climbing `(Total Bobot: 25 Poin)`

---

SMK TelkomBaru memiliki **8 mata pelajaran** yang harus dijadwalkan dalam satu semester. Tersedia **3 ruang kelas** dan **4 slot waktu** dalam seminggu.

**Syarat:**
- Tidak boleh ada dua mata pelajaran yang dijadwalkan pada **ruangan dan waktu yang sama**.
- Setiap mata pelajaran dijadwalkan tepat pada satu ruangan dan satu waktu.

**Final State:** Seluruh mata pelajaran terjadwal tanpa bentrok.

**Data:**
| Mata Pelajaran | Ruang Kelas | Waktu |
|---|---|---|
| Jaringan Komputer | Lab 1 | Senin 08:00 |
| Basis Data | Lab 2 | Senin 10:00 |
| Pemrograman Web | Lab 3 | Selasa 08:00 |
| Kecerdasan Artifisial | | Selasa 10:00 |
| Keamanan Siber | | |
| Sistem Operasi | | |
| Algoritma | | |
| Struktur Data | | |

```python
    # Import library
import random
import pandas as pd

def printJadwal(solusi):
        print(pd.DataFrame.from_dict(solusi, orient='index', columns=['Ruangan', 'Waktu']))
```

### 1a. Buat fungsi `initiateSolusiAwal()` `(Bobot: 5 Poin)`

Fungsi menerima tiga parameter: `mapel`, `ruangan`, `waktu`.
Kembalikan **dictionary** dengan key nama mata pelajaran dan value berupa **tuple** (ruangan, waktu) yang dipilih secara acak.

**Pro tips:** Gunakan `random.choice()`.

```python
# [1a] Fungsi inisiasi solusi awal
def initiateSolusiAwal(mapel, ruangan, waktu):
    solusi = {}  # Buat dictionary kosong
    for mk in mapel:  # Telusuri seluruh mata pelajaran
        solusi[mk] = (random.choice(ruangan), random.choice(waktu))  # Pilih ruangan & waktu acak
    return solusi
```

### 1b. Buat fungsi `hitungBentrok()` `(Bobot: 5 Poin)`

Fungsi menerima `solusi` (dictionary). Hitung dan kembalikan **jumlah jadwal yang bentrok** (pasangan ruangan+waktu yang sama digunakan lebih dari satu mata pelajaran).

```python
# [1b] Fungsi hitung bentrok
def hitungBentrok(solusi):
    jumlahBentrok = 0  # Inisiasi jumlah bentrok dengan 0
    jadwal = {}  # Buat dictionary kosong
    for mataKuliah, (ruangan, waktu) in solusi.items():
        key = (ruangan, waktu)
        if key in jadwal:  # Apakah slot sudah dipakai?
            jumlahBentrok += 1  # Increment bentrok
        else:
            jadwal[key] = mataKuliah  # Daftarkan slot ke jadwal
    return jumlahBentrok
```

### 1c. Buat fungsi `cariTetangga()` `(Bobot: 5 Poin)`

Fungsi menghasilkan seluruh kemungkinan tetangga dari solusi saat ini dengan cara **mencoba mengganti ruangan atau waktu** satu mata pelajaran per tetangga.

```python
# [1c] Fungsi cari tetangga
def cariTetangga(solusi, arrMapel, arrRuangan, arrWaktu):
    tetangga = []  # Inisiasi list kosong
    for pelajaran in arrMapel:  # Telusuri seluruh mata pelajaran
        currRuangan, currWaktu = solusi[pelajaran]
        # Ganti ruangan
        for ruangan in arrRuangan:
            if ruangan != currRuangan:  # Jika bukan ruangan saat ini
                tetanggaBaru = solusi.copy()
                tetanggaBaru[pelajaran] = (ruangan, currWaktu)
                tetangga.append(tetanggaBaru)

        # Ganti waktu
        for waktu in arrWaktu:
            if waktu != currWaktu:  # Jika bukan waktu saat ini
                tetanggaBaru = solusi.copy()
                tetanggaBaru[pelajaran] = (currRuangan, waktu)
                tetangga.append(tetanggaBaru)
    return tetangga
```

### 1d. Jalankan Algoritma Hill Climbing `(Bobot: 10 Poin)`

Gunakan fungsi-fungsi di atas untuk mencari jadwal tanpa bentrok.

```python
# [1d] Inisiasi data
mataPelajaran = ["Jaringan Komputer", "Basis Data", "Pemrograman Web", 
                 "Kecerdasan Artifisial", "Keamanan Siber",
                 "Sistem Operasi", "Algoritma", "Struktur Data"]  # 8 mata pelajaran sesuai soal
ruangKelas    = ["Lab 1","Lab 2", "Lab 3"]  # 3 ruang kelas: Lab 1, Lab 2, Lab 3
waktu         = ["Senin 08:00", "Senin 10:00", "Selasa 08:00", "Selasa 10:00"]  # 4 slot waktu sesuai soal
```

```python
# Buat solusi awal
solusi    = initiateSolusiAwal(mataPelajaran, ruangKelas, waktu)  # Panggil initiateSolusiAwal()
jumBentrok = hitungBentrok(solusi)  # Panggil hitungBentrok()
iterasi   = 0

print(f"Iterasi: {iterasi}\nJumlah Bentrok: {jumBentrok}\nSolusi Awal:")
printJadwal(solusi)

# Loop Hill Climbing
while True:
    tetangga = cariTetangga(solusi, mataPelajaran, ruangKelas, waktu)  # Cari tetangga
    jumlahBentrokTetangga = []  # List kosong
    for solusiTetangga in tetangga:
        bentrokTetangga = hitungBentrok(solusiTetangga)  # Hitung bentrok tetangga
        jumlahBentrokTetangga.append((solusiTetangga, bentrokTetangga))

    jumlahBentrokTetangga.sort(key=lambda x: x[1])
    bestTetangga, bestBentrok = jumlahBentrokTetangga[0]  # Ambil yang terbaik

    if bestBentrok < jumBentrok:  # Jika tetangga lebih baik
        solusi    = bestTetangga
        jumBentrok = bestBentrok
        iterasi  += 1
        print(f"\nIterasi: {iterasi}\nJumlah Bentrok: {jumBentrok}\nSolusi:")
        printJadwal(solusi)
        print("-" * 50)
        if bestBentrok == 0:  # Jika bentrok sudah 0
            print("Solusi terbaik ditemukan!")
            break
    else:
        print("Tidak ada tetangga yang lebih baik. Algoritma berhenti.")
        break

print("\nSolusi Akhir:")
printJadwal(solusi)
print("Jumlah bentrok:", jumBentrok)
```

## Soal 2: Implementasi Simulated Annealing `(Total Bobot: 25 Poin)`

---

**EkspedisiNusantara** memiliki 1 kantor pusat dan 8 lokasi pelanggan yang harus dikunjungi. Setiap kendaraan memiliki kapasitas maksimum. Minimumkan total jarak tempuh.

**Data Perusahaan:**
- Kantor pusat: koordinat `(40, 40)`
- Jumlah kendaraan: `3`
- Kapasitas tiap kendaraan: `80`

**Data Pelanggan:**
| ID | X | Y | Kapasitas |
|---|---|---|---|
| Cust-01 | 10 | 15 | 10 |
| Cust-02 | 20 | 25 | 15 |
| Cust-03 | 30 | 10 | 20 |
| Cust-04 | 15 | 35 | 25 |
| Cust-05 | 50 | 20 | 30 |
| Cust-06 | 55 | 45 | 10 |
| Cust-07 | 60 | 30 | 15 |
| Cust-08 | 70 | 50 | 20 |

```python
import random
import math
import matplotlib.pyplot as plt
```

### 2a. Fungsi `jarakEuclidean(a, b)` `(Bobot: 3 Poin)`

Hitung jarak Euclidean antara dua titik koordinat `a` dan `b` (masing-masing berupa tuple).

$$d = \sqrt{(x_2-x_1)^2 + (y_2-y_1)^2}$$

```python
# [2a] Jarak Euclidean
def jarakEuclidean(a, b):
    return math.sqrt((a[0] - b[0])**2 + (a[1] - b[1])**2)  # Hitung dengan rumus Euclidean, gunakan math.sqrt()
```

### 2b. Fungsi `cariPelanggan(customerID, arrCust)` `(Bobot: 2 Poin)`

Kembalikan data pelanggan (tuple) dari list `arrCust` berdasarkan `customerID`.

```python
# [2b] Cari pelanggan berdasarkan ID
def cariPelanggan(customerID, arrCust):
    for cust in arrCust:
        if cust[0] == customerID:
            return cust
    return None
```

### 2c. Fungsi `totalJarak(routes, pusat, arrCust)` `(Bobot: 5 Poin)`

Hitung total jarak seluruh kendaraan (berangkat dari pusat, kunjungi pelanggan, kembali ke pusat).

```python
# [2c] Total jarak
def totalJarak(routes, pusat, arrCust):
    distance = 0
    for route in routes:
        if len(route) == 0:
            continue  # skip jika kosong
        prev = cariPelanggan(route[0], arrCust)  # Mulai dari pusat
        for customerID in route:
            customer = cariPelanggan(customerID, arrCust)
            if customer is None:
                continue
            current = (customer[1], customer[2])
            distance += jarakEuclidean((prev[1], prev[2]), current)
            prev = customer
        distance += jarakEuclidean(pusat, (prev[1], prev[2]))  # Kembali ke pusat
    return distance
```

### 2d. Fungsi `validateRoute(routes, arrCust, kapasitasMax)` `(Bobot: 3 Poin)`

Kembalikan `True` jika tidak ada kendaraan yang melebihi kapasitas maksimum, `False` jika ada.

```python
# [2d] Validasi kapasitas
def validateRoute(routes, arrCust, kapasitasMax):
    for route in routes:
        total = 0
        for customerID in route:
            customer = cariPelanggan(customerID, arrCust)
            if customer is not None:
                total += customer[3]
        if total > kapasitasMax:
            return False
    return True
```

### 2e. Fungsi `inisiasiSolusiAwal(jumlahKendaraan, arrCust, kapasitasMaksimum)` `(Bobot: 4 Poin)`

Buat rute awal secara acak dengan mempertimbangkan kapasitas kendaraan.

```python
# [2e] Inisiasi solusi awal
def inisiasiSolusiAwal(jumlahKendaraan, arrCust, kapasitasMaksimum):
    routes = []
    for i in range(jumlahKendaraan):
        routes.append([])
    shuffleCust = []
    for cust in arrCust:
        shuffleCust.append(cust[0])
    random.shuffle(shuffleCust)
    for customerID in shuffleCust:
        customer = cariPelanggan(customerID, arrCust)
        placed = False
        for route in routes:
            total = sum(next(c for c in arrCust if c[0] == cid)[3] for cid in route)
            if total + customer[3] <= kapasitasMaksimum:
                route.append(customerID)
                placed = True
                break
        if not placed:
            routes[0].append(customerID)
    return routes
```

### 2f. Fungsi `cariTetangga(routes, arrCust, kapasitasMax)` `(Bobot: 4 Poin)`

Buat solusi tetangga dengan menukar satu pelanggan antara dua rute berbeda secara acak.

```python
# [2f] Cari tetangga
def cariTetangga(routes, arrCust, kapasitasMax):
    new_routes = []
    for route in routes:
        new_routes.append(route.copy())
    route1, route2 = random.sample(range(len(new_routes)), 2)
    if len(new_routes[route1]) == 0 or len(new_routes[route2]) == 0:
        return new_routes
    cust1 = random.choice(new_routes[route1])
    cust2 = random.choice(new_routes[route2])
    new_routes[route1].remove(cust1)
    new_routes[route2].remove(cust2)
    new_routes[route1].append(cust2)
    new_routes[route2].append(cust1)
    if validateRoute(new_routes, arrCust, kapasitasMax):
        return new_routes
    else:
        return routes
```

### 2g. Algoritma Simulated Annealing & Main Program `(Bobot: 4 Poin)`

Implementasikan fungsi `simulated_annealing()` dan jalankan program utama.

```python
# [2g] Simulated Annealing
def simulated_annealing(jumlahKendaraan, arrCust, kapasitasMaksimum, pusat):
    currentSolution = inisiasiSolusiAwal(jumlahKendaraan, arrCust, kapasitasMaksimum)
    currentCost = totalJarak(currentSolution, pusat, arrCust)
    bestSolution = currentSolution
    bestCost = currentCost
    T = 1000   # Temperatur awal (misal: 1000)
    T_min = 1
    alpha = 0.95   # Faktor pendinginan (misal: 0.95)
    while T > T_min:
        for _ in range(100):
            newSolution = cariTetangga(currentSolution, arrCust, kapasitasMaksimum)
            newCost = totalJarak(newSolution, pusat, arrCust)
            delta = newCost - currentCost
            if delta < 0 or random.uniform(0, 1) < math.exp(-delta / T):
                currentSolution = newSolution
                currentCost = newCost
                if currentCost < bestCost:
                    bestSolution = currentSolution
                    bestCost = currentCost
        T *= alpha
    return bestSolution, bestCost

# Data
kantorPusat = (40, 40)  # Koordinat kantor pusat
arrCust = [
    ("Cust-01", 10, 15, 10),
    ("Cust-02", 20, 25, 15),
    ("Cust-03", 30, 10, 20),
    ("Cust-04", 15, 35, 25),
    ("Cust-05", 50, 20, 30),
    ("Cust-06", 55, 45, 10),
    ("Cust-07", 60, 30, 15),
    ("Cust-08", 70, 50, 20),
]  # List tupple (ID, X, Y, Kapasitas)
jumlahKendaraan = 3
kapasitasMax = 80

# Jalankan SA
bestRoutes, bestDistance = simulated_annealing(jumlahKendaraan, arrCust, kapasitasMax, kantorPusat)

print("Rute Terbaik:")
for i in range(jumlahKendaraan):
    print(f"Kendaraan {i+1}:", end=" kantor pusat -> ")
    for j in range(len(bestRoutes[i])):
        print(bestRoutes[i][j], end=" -> ")
    print("kantor pusat")
print("\nJarak Terbaik:", bestDistance)
```

## Soal 3: Implementasi Greedy Best-First Search `(Total Bobot: 15 Poin)`

---

Di sebuah **rumah sakit**, terdapat beberapa ruangan yang terhubung dengan koridor. Seorang pasien ingin menuju **Ruang ICU** dari **Pintu Masuk** menggunakan jalur dengan estimasi jarak terpendek.

- `Pintu Masuk` ke `Apotek` -> 0
- `Lobi` ke `Poli Umum` -> 0
- `Lobi` ke `Radiologi` -> 0
- `Apotek` ke `Radiologi` -> 0
- `Poli Umum` ke `Ruang ICU` -> 0
- `Radiologi` ke `Ruang ICU` -> 0

*(Pada Greedy BFS, bobot edge tidak digunakan, sehingga semua bernilai 0)*

**Nilai Heuristik (estimasi jarak ke Ruang ICU):**
| Ruangan | Heuristik |
|---|---|
| Pintu Masuk | 500 |
| Lobi | 350 |
| Apotek | 400 |
| Poli Umum | 150 |
| Radiologi | 200 |
| Ruang ICU | 0 |

```python
import networkx as nx
import matplotlib.pyplot as plt

# Posisi node untuk visualisasi (!! Tidak usah dimodifikasi !!)
pos = {
    'Pintu Masuk' : (0, 0),
    'Lobi'        : (2, 2),
    'Apotek'      : (2, -2),
    'Poli Umum'   : (4, 3),
    'Radiologi'   : (4, 0),
    'Ruang ICU'   : (6, 1.5)
}

# Fungsi pendukung untuk mencetak graf (!! Tidak usah dimodifikasi !!)
def show_graph(G, pos=None, title=''):
    if pos is None:
        pos = nx.spring_layout(G)
    nx.draw(G, pos, with_labels=True, node_color='red', node_size=3000,
            font_color='white', font_weight='bold', font_size=9, width=4)
    edge_labels = nx.get_edge_attributes(G, 'weight')
    nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels,
                                  font_color='blue', font_size=12)
    plt.margins(0.2)
    plt.title(title)
    plt.show()
```

### 3a. Inisialisasi Graf & Tambahkan Node `(Bobot: 3 Poin)`

```python
# [3a] Inisialisasi graf tidak berarah
rumah_sakit = nx.Graph()

# Daftar ruangan
nodes = ["Pintu Masuk", "Lobi", "Apotek", "Poli Umum", "Radiologi","Ruang ICU"]  # Isi dengan nama-nama ruangan

# Tambahkan node
rumah_sakit.add_nodes_from(nodes)

# Tampilkan graf
show_graph(rumah_sakit, pos=pos, title="Graf Rumah Sakit - Node")
```

### 3c. Fungsi Heuristik `(Bobot: 4 Poin)`

Buat fungsi `heuristable(a, b)` berdasarkan tabel nilai heuristik di atas.

```python
# [3c] Fungsi heuristik
def heuristable(a, b):
    if a == "Pintu Masuk":  # Pintu Masuk
        return 500
    elif a == "Lobi":  # Lobi
        return 350
    elif a == "Apotek":
        return 400
    elif a == "Poli Umum":
        return 150
    elif a == "Radiologi":
        return 200
    elif a == "Ruang ICU":
        return 0
    # Lanjutkan untuk semua ruangan
```

### 3d. Cari Jalur dengan Greedy BFS `(Bobot: 5 Poin)`

Gunakan `nx.astar_path()` dengan heuristik yang sudah dibuat dan `weight='weight'`.

## Soal 6: Pengenalan Library Scikit-Fuzzy `(Total Bobot: 15 Poin)`

---

Anda diminta membangun **Sistem Fuzzy** untuk menentukan **Tingkat Layanan** sebuah restoran berdasarkan dua faktor:

- **Kualitas Makanan** (`food_quality`): skala **0–10**
- **Kecepatan Pelayanan** (`service_speed`): skala **0–10**

Output: **Tip** (persen tip yang diberikan), rentang **0–30**.

### Input (Antecedents)

**food_quality** (rentang [0, 10]):
- `bad`: Triangular `[0, 0, 5]`
- `average`: Triangular `[3, 5, 8]`
- `good`: Triangular `[6, 10, 10]`

**service_speed** (rentang [0, 10]):
- `slow`: Triangular `[0, 0, 5]`
- `moderate`: Triangular `[3, 5, 7]`
- `fast`: Triangular `[5, 10, 10]`

### Output (Consequent)

**tip** (rentang [0, 30]):
- `low`: Triangular `[0, 0, 15]`
- `medium`: Triangular `[10, 15, 20]`
- `high`: Triangular `[15, 30, 30]`

### Aturan Fuzzy
1. Jika `food_quality` **good** DAN `service_speed` **fast**, maka `tip` **high**.
2. Jika `food_quality` **average**, maka `tip` **medium**.
3. Jika `food_quality` **bad** ATAU `service_speed` **slow**, maka `tip` **low**.

**Simulasikan** dengan: `food_quality = 7`, `service_speed = 8`

```python
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import numpy as np
import matplotlib.pyplot as plt
```

### 6a. Inisialisasi Variabel Fuzzy `(Bobot: 3 Poin)`

```python
# [6a] Variabel input
food_quality   = ctrl.Antecedent(np.arange(0, 11, 1), 'food_quality')
service_speed  = ctrl.Antecedent(np.arange(0, 11, 1), 'service_speed')

# Variabel output
tip = ctrl.Consequent(np.arange(0, 31, 1), 'tip')
```

### 6b. Fungsi Keanggotaan `(Bobot: 4 Poin)`

Definisikan fungsi keanggotaan triangular untuk seluruh variabel dan tampilkan grafnya.

```python
# [6b] Keanggotaan food_quality
food_quality['bad']     = fuzz.trimf(food_quality.universe, [0, 0, 5])
food_quality['average'] = fuzz.trimf(food_quality.universe, [3, 5, 8])
food_quality['good']    = fuzz.trimf(food_quality.universe, [6, 10, 10])
food_quality.view()
plt.title("Keanggotaan Food Quality")

# Keanggotaan service_speed
service_speed['slow']     = fuzz.trimf(service_speed.universe, [0, 0, 5])
service_speed['moderate'] = fuzz.trimf(service_speed.universe, [3, 5, 7])
service_speed['fast']     = fuzz.trimf(service_speed.universe, [5, 10, 10])
service_speed.view()
plt.title("Keanggotaan Service Speed")

# Keanggotaan tip
tip['low']    = fuzz.trimf(tip.universe, [0, 0, 15])
tip['medium'] = fuzz.trimf(tip.universe, [10, 15, 20])
tip['high']   = fuzz.trimf(tip.universe, [15, 30, 30])
tip.view()
plt.title("Keanggotaan Tip")
```

### 6c. Aturan Fuzzy `(Bobot: 3 Poin)`

```python
# [6c] Aturan fuzzy
rule1 = ctrl.Rule(food_quality['good'] & service_speed['fast'], tip['high'])
rule2 = ctrl.Rule(food_quality['average'], tip['medium'])
rule3 = ctrl.Rule(food_quality['bad'] & service_speed['slow'], tip['low'])
```

### 6d. Simulasi & Hasil `(Bobot: 5 Poin)`

```python
# [6d] Sistem kontrol dan simulasi
tip_ctrl       = ctrl.ControlSystem([rule1, rule2, rule3])
tip_simulation = ctrl.ControlSystemSimulation(tip_ctrl)

# Input
food_q = 7  # food_quality = 7
service_s = 8  # service_speed = 8

# Hitung
tip_simulation.input['food_quality'] = food_q
tip_simulation.input['service_speed'] = service_s
tip_simulation.compute()

# Tampilkan hasil
print(f"Tip yang disarankan: {tip_simulation.output['tip']:.2f}%")
tip.view(sim=tip_simulation)  # Visualisasi
```
