```python
import random 
import math

# Menghitung jarak 2 titik dengan menggunakan rumus euclidean 
def distance(city1, city2): 
    return math.sqrt((city1[0] - city2[0]) ** 2 + (city1[1] - city2[1]) ** 2) 
 
# Menjumlah seluruh jarak antar titik pada rute tour 
def total_distance(tour, cities): 
    dist = 0 
    for i in range(len(tour)): 
        dist += distance(cities[tour[i]], cities[tour[(i + 1) % len(tour)]]) 
    return dist

def total_distance_mtsp(tour, split, cities):
    tour1 = tour[:split]
    tour2 = tour[split:]
    return total_distance(tour1, cities) + total_distance(tour2, cities)

# Random Swap dua titik dalam rute
def random_swap(tour, split):
    new_tour = tour [:]
    new_split = split

    if random.random() < 0.5:
        # Swap
        i, j = random.sample(range(len(tour)), 2)
        new_tour[i], new_tour[j] = new_tour[j], new_tour[i]
    else:
        new_split = random.randint(1, len(tour)-1)
    
    return new_tour, new_split

     
# Implementasi Algoritma SA
def simulated_annealing(cities, initial_temp, cooling_rate, stopping_temp):
    tour = list(range(len(cities)))
    random.shuffle(tour)
    split = len(cities) // 2

    current_distance = total_distance_mtsp(tour, split, cities)
    best_tour, best_split = tour[:], split
    best_distance = current_distance

    temp = initial_temp
    while temp > stopping_temp:
        new_tour, new_split = random_swap(tour, split)
        new_distance = total_distance_mtsp(new_tour, new_split, cities)

        if new_distance < current_distance:
            tour, split = new_tour[:], new_split
            current_distance = new_distance
        else:
            # Menghitung probabilitas MC
            if random.random() < math.exp((current_distance - new_distance)/temp):
                tour, split = new_tour[:], new_split
                current_distance = new_distance

        if current_distance < best_distance:
            best_tour, best_split, best_distance = tour[:], split, current_distance

        temp *= cooling_rate

    return best_tour, best_split, best_distance

cities = [ 
    (10, 20), (35, 15), (2, 40), (25, 30), 
    (15, 25), (40, 35), (20, 40), (50, 10) 
]

# Parameter SA
init = 1000
cooling = 0.995
stop = 1

best_tour, best_split, best_distance = simulated_annealing(cities, init, cooling, stop)

print("Rute Salesman 1:", best_tour[:best_split])
print("Rute Salesman 2:", best_tour[best_split:])
print("Total jarak terbaik:", best_distance)
```
